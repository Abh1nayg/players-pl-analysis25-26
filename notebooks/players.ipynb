checking k means clusters

```python
from sklearn.cluster import KMeans
```

```python
X=sclInput
X
```

Importing silhouette 

```python
from sklearn.metrics import silhouette_score
```

```python
inertias = []
silhouette_scores = []

for i in range(2, 11): 
    kms = KMeans(n_clusters=i, random_state=42)
    kms.fit(dfMm[sclInput])
    
    inertia = kms.inertia_
    sil_score = silhouette_score(dfMm[sclInput], kms.labels_)
    
    inertias.append(inertia)
    silhouette_scores.append(sil_score)
    
    # print(f"\n--- K={i} ---")
    # print("Cluster labels:", kms.labels_)
    # print("Cluster centroids:\n", kms.cluster_centers_)
    print(f"K={i} | Inertia: {inertia:.2f} | Silhouette Score: {sil_score:.4f}")
```

```python
import matplotlib.pyplot as plt

# Create the line graph for Inertia (dfMm)
plt.figure(figsize=(8, 5))
plt.plot(range(2, 11), inertias, marker='o', linestyle='-', color='b')

# Add labels and formatting
plt.title('Elbow Method For Optimal k (MinMax Scaled Data)')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.xticks(range(2, 11))
plt.grid(True, linestyle='--', alpha=0.7)

# Display the plot
plt.show()
```

```python
inertias = []
silhouette_scores = []

for i in range(2, 11): 
    kms = KMeans(n_clusters=i, random_state=42)
    kms.fit(dfZ[sclInput])
    
    inertia = kms.inertia_
    sil_score = silhouette_score(dfZ[sclInput], kms.labels_)
    
    inertias.append(inertia)
    silhouette_scores.append(sil_score)
    
    # print(f"\n--- K={i} ---")
    # print("Cluster labels:", kms.labels_)
    # print("Cluster centroids:\n", kms.cluster_centers_)
    print(f"K={i} | Inertia: {inertia:.2f} | Silhouette Score: {sil_score:.4f}")
```
